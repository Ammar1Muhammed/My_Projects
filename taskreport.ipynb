```python
import numpy as np
                    #المبيعات
sales = np.array([
    [10, 15, 8],   # يوم 1
    [12, 20, 5],   # يوم 2
    [30, 10, 2],   # يوم 3
    [8, 12, 10],   # يوم 4
    [15, 18, 7]    # يوم 5
])
                                           #مجموع المبيعات من كل نوع
total_sales_of_typs= sales.sum(axis=0)
print("Total sales of typs\n---------")
print("sumsunge=",total_sales_of_typs[0])
print("hawawy=",total_sales_of_typs[1])
print("iphon=",total_sales_of_typs[2],"\n---------")
                                           #مجموع المبيعات في كل يوم
total_sales_of_days = sales.sum(axis=1)
print("Sum of days\n------")
print("day1=",total_sales_of_days[0])
print("day2=",total_sales_of_days[1])
print("day3=",total_sales_of_days[2])
                                          #مجموع المبيعات كلها
total=sales.sum()
print(f"------\nTotal: {total}")
                                          #اقل مبيعات في كله
less=sales.min()
print(f"-------\nLess sales: {less}")
                                          #اكبر مبيعات في كله
most=sales.max()
print(f"-------\nMost sales: {most}")
                                          #اقل مبيعات في النوع
less_sales_typs=sales.min(axis=0)
print("-------\nLess sales of typs\n-------")
print("sumsunge=",less_sales_typs[0])
print("hawawy=",less_sales_typs[1])
print("iphon=",less_sales_typs[2],"\n---------")
                                         #اقل مبيعات في كل يوم
less_sales_days=sales.min(axis=1)
print("-------\nLess sales of days\n-------")
print("day1=",less_sales_days[0])
print("day2=",less_sales_days[1])
print("day3=",less_sales_days[2],"\n---------")
                                        #اكبر مبيعات من كل نوع
most_sales_of_typs = sales.max(axis=0)
print("Most sales of typs\n---------")
print("sumsunge=",most_sales_of_typs[0])
print("hawawy=",most_sales_of_typs[1])
print("iphon=",most_sales_of_typs[2],"\n---------")
                                       #اكبر مبيعات في كل يوم
most_sales_days=sales.max(axis=1)
print("-------\nMost sales of days\n-------")
print("day1=",most_sales_days[0])
print("day2=",most_sales_days[1])
print("day3=",most_sales_days[2],"\n---------")
                                       #متوسط المبيعات في كل نوع
Average_sales_of_typs = sales.mean(axis=0)
print("Average sales of typs\n---------")
print("sumsunge=",Average_sales_of_typs[0])
print("hawawy=",Average_sales_of_typs[1])
print("iphon=",Average_sales_of_typs[2],"\n---------")
                                        #متوسط المبيعات في كل يوم
Average_sales_days=sales.mean(axis=1)
print("-------\nAverage sales of days\n-------")
print("day1=",Average_sales_days[0])
print("day2=",Average_sales_days[1])
print("day3=",Average_sales_days[2],"\n---------")
                                        #متوسط المبيعات في كله
Average=sales.mean()
print(f"Average sales= {Average}\n--------")
                        #اليوم التالت
print("Day3\n------")
                        #مجموع المبيعات في اليوم التالت
total_sale_of_day3=sales[2].sum()
print(f"Total sale = {total_sale_of_day3}")
                        #اكبر مبيعات في اليوم التالت
most_sale_day3=sales[2].max()
print(f"Most sale = {most_sale_day3} ")
                        #اقل مبيعات في اليوم التالت
less_sale_day3=sales[2].min()
print(f"Less sale = {less_sale_day3} ")
                        #متوسط المبيعات في اليوم التالت
Average_day3=sales[2].mean()
print(f"Average = {Average_day3}") 





```
