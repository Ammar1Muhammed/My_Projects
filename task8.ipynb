```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


df = pd.read_csv(
    "e_commerce.csv",
    dtype={
        "order_id": np.int16,
        "quantity": np.int8,
        "shipping_days": np.int8,
    },
    true_values=['Yes'],
    false_values=['No'],
    index_col=0,
)
df['order_date'] = pd.to_datetime(df['order_date'])
df.head()
```

```python
unit_price_after_discount = np.round(df['unit_price'] * (1 - df['discount_pct']), 2)

df.insert(7, "unit_price_after_discount", unit_price_after_discount)
df.insert(8, "total_price_after_discount", unit_price_after_discount * df['quantity'])
df.head()
```

```python
pd.DataFrame(df[df["product_name"]== df["product_name"]].groupby(["product_name"])["shipping_days"].sum().sort_values(ascending=False).head(10))
```

```python
pd.DataFrame(df[df["country"]== df["country"]].groupby(["country"])["quantity"].sum().sort_values(ascending=False).head(10))
```

```python
x=df[["unit_price","discount_pct","quantity","shipping_days"]]
y=df["return_status"]
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
model_1=LogisticRegression()
model_1.fit(x_train,y_train)
f"score:{model_1.score(x_test,y_test)*100:.2f}%"
```

```python
prediction_1 = model_1.predict(x_test)
result=pd.DataFrame({"unit_price": x_test["unit_price"],"discount_pct":x_test["discount_pct"],"quantity":x_test["quantity"],"shipping_days":x_test["shipping_days"],"return_status":y_test, "Predicted_return_status": prediction_1})
result.head()
```

```python
prediction_1=model_1.predict([[30,0.5,3,2]])
f"Predicted return status for unit price 30, discount percentage 0.5, quantity 3 and shipping days 2: {prediction_1}"
```
