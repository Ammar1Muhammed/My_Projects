```python
import pandas as pd
import numpy as np

sweets = pd.DataFrame({
    "كنافة":np.array([12,14,9,20,23]),
    "بسبوسة":np.array([15,14,23,25,22]),
    "زلابيا":np.array([4,7,10,5,7]),
    "بقلاوة":np.array([10,14,11,22,23]),
    "ام علي":np.array([2,6,8,3,7])
},index=["total ","Day 2","Day 3","Day 4","Day 5"])
sweets
```
